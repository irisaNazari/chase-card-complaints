```python
import duckdb

con = duckdb.connect()
con.sql("DESCRIBE SELECT * FROM '../data/raw/CCDB_Export_21_August_2026.csv'").df()
```

```python
con.sql("""
    SELECT Product, COUNT(*) AS complaints
    FROM '../data/raw/CCDB_Export_21_August_2026.csv'
    GROUP BY Product
    ORDER BY complaints DESC
""").df()
```

```python
con.sql("""
    SELECT "Company response to consumer", COUNT(*) AS complaints
    FROM '../data/raw/CCDB_Export_21_August_2026.csv'
    WHERE Product = 'Credit card'
    GROUP BY "Company response to consumer"
    ORDER BY complaints DESC
""").df()
```

```python
con.sql("""
    SELECT "Sub-product", COUNT(*) AS complaints
    FROM '../data/raw/CCDB_Export_2_May_2018_through_April_2021.csv'
    WHERE Product = 'Credit card or prepaid card'
    GROUP BY "Sub-product"
    ORDER BY complaints DESC
""").df()
```

```python
con.sql("""
    SELECT "Sub-product", COUNT(*) AS complaints
    FROM '../data/raw/CCDB_Export_21_August_2026.csv'
    WHERE Product = 'Credit card'
    GROUP BY "Sub-product"
    ORDER BY complaints DESC
""").df()
```

```python
con.sql("""
    SELECT Company, COUNT(*) AS complaints
    FROM '../data/raw/CCDB_Export_2_May_2018_through_April_2021.csv'
    WHERE "Sub-product" IN ('General-purpose credit card or charge card', 'Store credit card')
    GROUP BY Company
    ORDER BY complaints DESC
    LIMIT 15
""").df()
```

```python
con.sql("""
    SELECT Company, COUNT(*) AS complaints
    FROM '../data/raw/CCDB_Export_2_May_2018_through_April_2021.csv'
    WHERE "Sub-product" IN ('General-purpose credit card or charge card', 'Store credit card')
      AND (Company ILIKE '%chase%' OR Company ILIKE '%jpmorgan%')
    GROUP BY Company
    ORDER BY complaints DESC
""").df()
```

```python
con.sql("""
    SELECT Issue, COUNT(*) AS complaints
    FROM '../data/raw/CCDB_Export_2_May_2018_through_April_2021.csv'
    WHERE "Sub-product" IN ('General-purpose credit card or charge card', 'Store credit card')
      AND Company = 'JPMORGAN CHASE & CO.'
    GROUP BY Issue
    ORDER BY complaints DESC
    LIMIT 10
""").df()
```

```python
con.sql("""
    SELECT Issue, COUNT(*) AS complaints
    FROM '../data/raw/CCDB_Export_21_August_2026.csv'
    WHERE "Sub-product" IN ('General-purpose credit card or charge card', 'Store credit card')
      AND Company = 'JPMORGAN CHASE & CO.'
    GROUP BY Issue
    ORDER BY complaints DESC
    LIMIT 10
""").df()
```

```python
con.sql("""
    SELECT
        Issue,
        "Sub-issue",
        COUNT(*) AS complaints
    FROM '../data/raw/CCDB_Export_21_August_2026.csv'
    WHERE Company = 'JPMORGAN CHASE & CO.'
    GROUP BY Issue, "Sub-issue"
    ORDER BY complaints DESC
""").df()
```
